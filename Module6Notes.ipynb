```python
def roastMarshmallow():
    print("Start a Bonfire")
    print("Put a marshmallow on a stick")
    print("Roast marshmallow on top of fire until golden")
    print("Take marshmallow off stick and eat it")

for i in range(2):
    roastMarshmallow()
```

```python
def sum_two_number(x,y):
    print(x+y)

sum_two_number(1,2)
sum_two_number(20,21)
sum_two_number(100,500)
```
