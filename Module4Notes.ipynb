```python
family_toppings = {
    "Charlie" : "anchovies",
    "Rose" : "basil",
    "Nathan" : "Cheese",
    "Judy" : "Garlic"
}

family_toppings["barry"] = "Berries"
family_toppings["Charlie"] = "Carolina Reaper"
print(family_toppings["barry"])
print(len(family_toppings))
print(family_toppings)

family_toppings["Judy"].title()
print(family_toppings)

family_toppings["Judy"] = family_toppings["Judy"].title()
print(family_toppings.get("Aliyah", "Aliyah is not in the family"))
print(family_toppings.get("Nathan", "Nathan is not in the family"))

for k,v in family_toppings.items():
    print(k)
    print(v)

for name in family_toppings.keys():
    print(name)

for topping in family_toppings.values():
    print(topping)
```

#### Lists allow us to store many values in one place and we access them by theif index number in the list. Dictionaries allow us to store things by key and we access the value by referencing the key. It is important to keep track of what data type we are using at any given time. For example, in the example above, censu is a dictionary. census["household2"] is also a dictionary. census["household2"]["income"] is a list and census["household2"]["income"][2] is an integer.

```python
color = ["Red", "Yellow", "Green"]
fav_colors = {
    "Grandma": "Red",
    "Tyler": "Blue",
    "Blake": "Purple"
}
for k in fav_colors.keys():
    print(k)
for v in fav_colors.values():
    print(v)
for k,v in fav_colors.items():
    print(k,v)

fav_colors.get("grandpa", "that key does not exist")

grouped_flavors = {
    "Cookies": {
        "Chocolate Chips": "Sweet",
        "Raisin": "Unami",
        "Sugar": "Sweet"
    },
    "Pizza": {
        "Pepperoni": "Savory",
        "Cheese": "Savory",
        "Mushrooms": "Unami"
    }
}
print(grouped_flavors["Cookies"]["Chocolate Chip"])
```
