```python
word_answer = "Golf"
word = input("Guess the word \n Hint, It is a sport that uses a stick.")
count = 0
while word != word_answer:
        print("Nope!")
        count = count + 1
        word = input("Guess the word \n Hint, It is a sport that uses a stick.")

print("You got it")
print(f"it took you {count} ammount of tries.")
```
