```python
words = ["Hello", "pizza", "school", "friendly", "words"]
print(words)
print(words[0])
print(words[4].upper())
print(words[-1].upper())

print(words[1].title())

print("One of my favorite works is:", words[3])
my_sentence = f"One of my favorite words is: {words[3]}"
print(my_sentence)

length = len(words)
print(length)
print(len(words))

sentence_2 = f"I love {words[1]} and {words[3]} but I hate {words[2]} my list of favorite words has {len(words)} items."
print(sentence_2)
```

### Every item in a list has a location starting at 0. We can also start from the end with -1. The location is referred to as an index. So the first item in a list is at index 0. We can use the items in the list as is appropriate according to their type. If the items are strings, we can call string methods on them. If they are numbers, we can do math with them. The way we access a particular item in a list is the name_of_list[index]. There is a len function that will tell us how long the list is.
