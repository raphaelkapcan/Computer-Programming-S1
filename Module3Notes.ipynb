```python
x = 5
y = 5
z = 6

# comparison operators to build one statement

print(x != y)
print(x != z)
print(x == y)
print(x == z)
print(x >= y)
print(x > y)
print(x <= y)

#logical operators to compare multiple statements that each were built with comparison operators

print(x!=y and x!=z)
print(x!=y or x!=z)
print(not(x!=y))
```

If there are multiple routes, we can include some elifs in between the if and else.

```python
clothing_stores = ["Walmart", "Pandabuy", "Nike", "Peter Millar", "Dior"]
fav_store = input("What is your favorite clothing store?")
has_store = False

for store in clothing_stores:
    if store.lower() == fav_store.lower():
        has_store = True
if has_store == True:
    print("We have that store")
else:
    print("We don't have that store")
```
